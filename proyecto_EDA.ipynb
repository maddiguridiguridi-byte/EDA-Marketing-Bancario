Por ultimo, analizaremos date:


```python
df_combinado['date'].head(10)
```

Vemos que las fechas siguen el formato día-mes-año, pero los meses están escritos en español.

Para poder analizar la evolución temporal, necesitamos que Python reconozca estos valores como fechas y no como simples textos.

```python
#Convertir date a formato fecha: 

#Primero, utilizaremos un diccionario para relacionar cada mes en español con su equivalente en ingles 
meses = {
    'enero': 'January',
    'febrero': 'February',
    'marzo': 'March',
    'abril': 'April',
    'mayo': 'May',
    'junio': 'June',
    'julio': 'July',
    'agosto': 'August',
    'septiembre': 'September',
    'octubre': 'October',
    'noviembre': 'November',
    'diciembre': 'December'
}
```

```python
#Aplicaremos las sustituciones de los meses al ingles: 

date_ingles = df_combinado['date'].replace(meses, regex=True)

date_ingles.head(10)
```

```python
#Ahora vamos a convertir a formato datetime:

date_convertida = pd.to_datetime(date_ingles, format='%d-%B-%Y', errors='coerce')

date_convertida.head(10)
```

```python
#Por ultimo, incorporaremos la conversion al DateFrame:

df_combinado['date'] = date_convertida

df_combinado['date'].dtype
```
